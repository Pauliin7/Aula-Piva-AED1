```python
def eh_par(n):
  if ( n % 2) == 0:
    return True
  else:
    return False
```

```python
n = int(input("Digite um numero: "))
if eh_par(n):
  print("O numero é par!")
else:
  print("O numero é impar")
```

```python
def calcular_imc(p, a):
  imc = p / (a ** 2)
  return imc
```

```python
def classificar_imc(imc):
  if (imc<18.5):
    return "Abaixo do peso!"
  elif (imc <24.9):
    return "Peso normal"
  elif (imc<29.9):
    return "Sobrepeso"
  else:
    return "Obesidade"
```

```python
p = float(input("Peso em kg"))
a = float(input("Altura em metros"))
imc = calcular_imc(p,a)
print(imc)
print(classificar_imc(imc))
```
