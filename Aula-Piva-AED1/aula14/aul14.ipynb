Relatorio Financeiro com Imposto opcional.

```python
def gerar_relatorio_financeiro(receita,despesa,imposto=0.15,mostrar_resumo=True):
  lucro_bruto = receita - despesa

  if lucro_bruto <= 0:
   return "Prejuizo registrado. Sem aplicação de imposto"

  imposto_aplicado = lucro_bruto * imposto
  lucro_liquido = lucro_bruto - imposto_aplicado

  if mostrar_resumo:
    print("Resumo do relatório financeiro")
    print(f"receita: R${receita}")
    print(f"despesa: R${despesa}")
    print(f"lucro bruto: R${lucro_bruto}")
    print(f"imposto aplicado: R${imposto_aplicado:.2f}")

  return lucro_liquido
```

Faca uma funcao que calcula o valor a ser cobrado pelos serviços de entrega de uma loja na internet. o custo é calculado tomando como base: o peso do produto = ate 500g R$15,00, até 2kg = R$43,00, acima de 2kg =R$103. O estado de destino: norte e nordeste=R$50,  centro-oeste=R$25 e sudeste = R$0. presente (é um presente), consideramos sempre que nao é presente = false. se for presente (true) adicione R$10

```python
# Mapeamento de abreviações de estado para regiões
ESTADO_PARA_REGIAO = {
    "AC": "Norte", "AP": "Norte", "AM": "Norte", "PA": "Norte", "RO": "Norte", "RR": "Norte", "TO": "Norte",
    "AL": "Nordeste", "BA": "Nordeste", "CE": "Nordeste", "MA": "Nordeste", "PB": "Nordeste", "PE": "Nordeste", "PI": "Nordeste", "RN": "Nordeste", "SE": "Nordeste",
    "GO": "Centro-Oeste", "MT": "Centro-Oeste", "MS": "Centro-Oeste", "DF": "Centro-Oeste",
    "ES": "Sudeste", "MG": "Sudeste", "RJ": "Sudeste", "SP": "Sudeste",
    "PR": "Sul", "SC": "Sul", "RS": "Sul" # Adicionado Sul para completude, embora não tenha regras de custo específicas ainda
}

def regiao(estado_abbr):
  """Retorna a região de um estado dada sua abreviação."""
  return ESTADO_PARA_REGIAO.get(estado_abbr.upper(), "Desconhecida")
```

```python
def frete(peso_produto, estado_destino_abbr, is_presente=False):
  custo_peso = 0

  if peso_produto <= 500:
    custo_peso = 15.00
  elif peso_produto <= 2000:
    custo_peso = 43.00
  else:
    custo_peso = 103.00

  custo_estado = 0

  # Usa a função auxiliar para obter a região
  regiao_destino = regiao(estado_destino_abbr) # Corrected function call to 'regiao'

  if regiao_destino == "Norte" or regiao_destino == "Nordeste":
    custo_estado = 50.00
  elif regiao_destino == "Centro-Oeste":
    custo_estado = 25.00
  elif regiao_destino == "Sudeste":
    custo_estado = 0.00
  else:
    print(f"Atenção: Região '{regiao_destino}' (do estado '{estado_destino_abbr}') não reconhecida para cálculo de frete, custo de entrega para estado pode ser zero.")

  custo_presente = 0

  if is_presente:
    custo_presente = 10.00

  custo_total = custo_peso + custo_estado + custo_presente
  return custo_total
```

### Pedra, Papel, Tesoura

```python
import random

def play_rock_paper_scissors():
    choices = ['pedra', 'papel', 'tesoura']
    player_score = 0
    computer_score = 0

    print("\n--- Jogo: Pedra, Papel, Tesoura ---")

    while True:
        player_choice = input("Escolha (pedra, papel, tesoura) ou 'sair' para encerrar: ").lower()

        if player_choice == 'sair':
            print(f"\nFim de jogo! Pontuação final: Você {player_score} x {computer_score} Computador")
            break

        if player_choice not in choices:
            print("Escolha inválida. Por favor, digite 'pedra', 'papel' ou 'tesoura'.")
            continue

        computer_choice = random.choice(choices)
        print(f"Você escolheu: {player_choice}")
        print(f"Computador escolheu: {computer_choice}")

        if player_choice == computer_choice:
            print("Empate!")
        elif (
            (player_choice == 'pedra' and computer_choice == 'tesoura') or
            (player_choice == 'papel' and computer_choice == 'pedra') or
            (player_choice == 'tesoura' and computer_choice == 'papel')
        ):
            print("Você ganhou!")
            player_score += 1
        else:
            print("Computador ganhou!")
            computer_score += 1

        print(f"Pontuação atual: Você {player_score} x {computer_score} Computador\n")

play_rock_paper_scissors()
```
